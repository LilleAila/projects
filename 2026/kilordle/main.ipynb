```python
import numpy as np
```

```python
words = np.loadtxt("words.csv", dtype=str)
```

```python
information_gains = np.zeros((26, 5), dtype=float)
n = len(words)

def char_idx(c):
    return ord(c) - ord("a")

for word in words:
    for i, char in enumerate(word):
        char_index = char_idx(char)
        information_gains[char_index, i] += 1

information_gains /= n
```

```python
used = np.zeros((26, 5), dtype=bool)
def use_word(word):
    for i, char in enumerate(word):
        used[char_idx(char), i] = True
```

```python
def information_gain(word):
    ig = 0.0
    for i, char in enumerate(word):
        c = char_idx(char)
        if used[c, i]:
            continue
        ig += information_gains[c, i]
    return ig
```

```python
guessed = []

while not np.all(used):
    best_word = max(words, key=information_gain)
    ig = information_gain(best_word)
    if ig == 0.0:
        break
    guessed.append(best_word)
    use_word(best_word)
```

```python
for i in guessed:
    print(i)
```
